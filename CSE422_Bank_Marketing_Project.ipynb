# CSE422 Lab Project â€“ Bank Marketing Dataset
## Binary Classification: Will the client subscribe to a term deposit?

```python
# ============================================================
# CSE422 Lab Project â€“ Bank Marketing Dataset
# Binary Classification: Will the client subscribe?
# ============================================================

# --- All Imports ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             confusion_matrix, roc_auc_score, roc_curve,
                             ConfusionMatrixDisplay)

print('âœ… All libraries imported successfully.')
```

## Section 1 â€“ Introduction

This project aims to predict whether a client of a Portuguese bank will subscribe to a term deposit product based on data collected during direct telemarketing campaigns. The ability to accurately predict subscription likelihood is valuable because it allows banks to optimize their marketing efforts, focusing resources on clients who are most likely to convert. Traditional mass telemarketing is expensive and often has low success rates; a machine-learning-driven targeting strategy can significantly reduce costs while improving campaign effectiveness.

We approach this as a **supervised binary classification** problem, training five different classifiers â€” K-Nearest Neighbors, Decision Tree, Logistic Regression, Naive Bayes, and a Neural Network (MLP) â€” and comparing their performance. Additionally, we apply **K-Means clustering** (unsupervised learning) to discover natural groupings in the data. The dataset contains 41,188 records with 20 input features covering client demographics, campaign details, and socio-economic indicators.

## Section 2 â€“ Dataset Description
### 2.1 Dataset Description

```python
# --- Load Dataset ---
df = pd.read_csv('Bank_Marketing.csv')
print('ðŸ“Š Dataset Shape:', df.shape)
print(f'   âžœ {df.shape[0]} data points, {df.shape[1]} features (including target)')
print()

# Identify feature types
num_cols = df.select_dtypes(include='number').columns.tolist()
cat_cols = df.select_dtypes(include='object').columns.tolist()

print(f'ðŸ”¢ Quantitative features ({len(num_cols)}): {num_cols}')
print(f'ðŸ”¤ Categorical features ({len(cat_cols)}): {cat_cols}')
print()
print('This is a BINARY CLASSIFICATION problem because the target variable y')
print('has exactly two classes: yes (subscribed) and no (did not subscribe).')
print()
print('âš ï¸ Categorical variables must be encoded to numerical values because')
print('   ML algorithms operate on numerical input only.')
print()
df.head()
```

```python
# --- Correlation Heatmap (after encoding categorical variables) ---
df_encoded = df.copy()
le_dict = {}
for col in df_encoded.select_dtypes(include='object').columns:
    le_dict[col] = LabelEncoder()
    df_encoded[col] = le_dict[col].fit_transform(df_encoded[col])

corr_matrix = df_encoded.corr()

plt.figure(figsize=(16, 12))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', annot_kws={'size': 7})
plt.title('Correlation Heatmap â€“ Bank Marketing Dataset', fontsize=14)
plt.tight_layout()
plt.show()

print('ðŸ“Š Observations:')
print('   âžœ duration has the strongest positive correlation with target y.')
print('   âžœ Economic indicators (euribor3m, emp.var.rate, nr.employed) are highly')
print('     inter-correlated, indicating multicollinearity.')
```

### 2.2 Imbalanced Dataset Check

```python
# --- Class Distribution ---
class_counts = df['y'].value_counts()
class_pct = df['y'].value_counts(normalize=True) * 100

print('ðŸ”´ Class Distribution:')
for cls in class_counts.index:
    print(f"   '{cls}': {class_counts[cls]} samples ({class_pct[cls]:.1f}%)")

plt.figure(figsize=(6, 4))
sns.countplot(x='y', data=df, palette='viridis')
plt.title('Distribution of Target Variable (y)', fontsize=13)
plt.xlabel('Subscribed to Term Deposit')
plt.ylabel('Count')
plt.tight_layout()
plt.show()

print('âš ï¸ The dataset is IMBALANCED (~88.7% no vs ~11.3% yes).')
print('   Accuracy alone is NOT a reliable metric; we must also examine')
print('   Precision, Recall, and AUC.')
```

### 2.3 Exploratory Data Analysis (EDA)

## Section 3 â€“ Dataset Pre-processing

```python
# -------------------------------------------------------
# Problem 1: Categorical Variables
# -------------------------------------------------------
print('ðŸ”´ Problem 1: Categorical variables cannot be used directly by ML models.')
print('   Solution : Apply LabelEncoder to convert each categorical column to integers.')
print()

cat_columns = df.select_dtypes(include='object').columns.tolist()
print(f'   Categorical columns to encode ({len(cat_columns)}): {cat_columns}')
print()

le_encoders = {}
for col in cat_columns:
    le_encoders[col] = LabelEncoder()
    df[col] = le_encoders[col].fit_transform(df[col])

print('âœ… All categorical columns encoded successfully.')
print(df.dtypes)
```

```python
# -------------------------------------------------------
# Problem 2: Missing / Null Values
# -------------------------------------------------------
print('ðŸ”´ Problem 2: Missing or null values can break model training.')
print()

null_counts = df.isnull().sum()
total_nulls = null_counts.sum()
print(f'   Total null values in dataset: {total_nulls}')
print()
print('âœ… The Bank Marketing dataset has ZERO missing values â€” no imputation needed.')
```

```python
# -------------------------------------------------------
# Problem 3: Feature Scaling
# -------------------------------------------------------
print('ðŸ”´ Problem 3: Features have different scales/magnitudes.')
print('   KNN is distance-based and Neural Networks use gradient descent â€”')
print('   both are sensitive to feature magnitude differences.')
print()
print('   Solution: Apply StandardScaler (zero mean, unit variance).')
print()

X = df.drop('y', axis=1)
y = df['y']

print('ðŸ“Š Before Scaling (sample statistics):')
print(X.describe().loc[['mean', 'std']].round(2).to_string())

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

print()
print('ðŸ“Š After Scaling (sample statistics):')
print(X_scaled.describe().loc[['mean', 'std']].round(2).to_string())
print()
print('âœ… Feature scaling complete.')
```

## Section 4 â€“ Dataset Splitting

```python
# --- Stratified Train/Test Split ---
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print('ðŸ“Š Dataset Split (80% Train / 20% Test â€” Stratified):')
print(f'   Train size: {len(X_train)}')
print(f'   Test  size: {len(X_test)}')
print(f'   Train class distribution:')
print(y_train.value_counts(normalize=True).round(4))
print(f'   Test  class distribution:')
print(y_test.value_counts(normalize=True).round(4))
print()
print('âœ… Stratified split preserves the 88.7/11.3 class ratio in both sets,')
```

## Section 5 â€“ Model Training & Testing
### 5.1 K-Nearest Neighbors (KNN)

```python
# --- KNN ---
knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(X_train, y_train)
print('âœ… KNN model trained (K=7).')
```

### 5.2 Decision Tree

```python
# --- Decision Tree ---
dt = DecisionTreeClassifier(max_depth=8, random_state=42)
dt.fit(X_train, y_train)
print('âœ… Decision Tree model trained (max_depth=8).')
```

### 5.3 Logistic Regression

```python
# --- Logistic Regression ---
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_train, y_train)
print('âœ… Logistic Regression model trained.')
```

### 5.4 Naive Bayes

```python
# --- Naive Bayes ---
nb = GaussianNB()
nb.fit(X_train, y_train)
print('âœ… Naive Bayes model trained.')
```

### 5.5 Neural Network (MLP) â€” MANDATORY

```python
# --- Neural Network (MLP) ---
mlp = MLPClassifier(hidden_layer_sizes=(128, 64), max_iter=300,
                    random_state=42, early_stopping=True,
                    activation='relu', solver='adam')
mlp.fit(X_train, y_train)
print('âœ… Neural Network (MLP) model trained (128-64 architecture).')
```

### 5.6 K-Means Clustering â€” Unsupervised

```python
# --- K-Means: Elbow Method ---
inertias = []
K_range = range(2, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertias.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertias, 'o-', color='crimson', linewidth=2)
plt.title('Elbow Method for Optimal K', fontsize=13)
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.xticks(list(K_range))
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Final model with K=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)

# Visualize with PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=cluster_labels, cmap='viridis', s=5, alpha=0.5)
plt.title('K-Means Clusters (PCA Projection)', fontsize=13)
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.colorbar(label='Cluster')
plt.tight_layout()
plt.show()

print('âœ… K-Means clustering complete (K=3).')
print('ðŸ“Š Note: Clusters do not perfectly align with binary labels since K-Means')
print('   is unsupervised â€” the PCA projection shows some overlap.')
```

## Section 6 â€“ Model Comparison Analysis

```python
# --- Collect Results for All Supervised Models ---
models_dict = {
    'KNN': knn,
    'Decision Tree': dt,
    'Logistic Regression': lr,
    'Naive Bayes': nb,
    'Neural Network': mlp
}

results = {}
for name, model in models_dict.items():
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    results[name] = {
        'accuracy':  accuracy_score(y_test, y_pred),
        'precision': precision_score(y_test, y_pred, zero_division=0),
        'recall':    recall_score(y_test, y_pred, zero_division=0),
        'auc':       roc_auc_score(y_test, y_prob),
        'y_pred':    y_pred,
        'y_prob':    y_prob
    }
    print(f'âœ… {name} evaluated.')

print()
print('ðŸ“Š All models evaluated successfully.')
```

### 6.1 Accuracy Bar Chart

```python
# --- 6.1 Accuracy Bar Chart ---
names = list(results.keys())
accs = [results[n]['accuracy'] for n in names]

plt.figure(figsize=(10, 5))
bars = plt.bar(names, accs, color=['#3498db','#2ecc71','#e74c3c','#f39c12','#9b59b6'],
               edgecolor='black')
for bar, acc in zip(bars, accs):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.002,
             f'{acc:.2%}', ha='center', va='bottom', fontweight='bold')
plt.ylim(0.80, 1.0)
plt.title('Model Accuracy Comparison', fontsize=13)
plt.ylabel('Accuracy')
plt.tight_layout()
plt.show()
```

### 6.2 Precision & Recall Grouped Bar Chart

```python
# --- 6.2 Precision & Recall ---
precs = [results[n]['precision'] for n in names]
recs  = [results[n]['recall'] for n in names]
x = np.arange(len(names))
w = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(x - w/2, precs, w, label='Precision', color='#2980b9', edgecolor='black')
ax.bar(x + w/2, recs,  w, label='Recall',    color='#e67e22', edgecolor='black')
ax.set_xticks(x)
ax.set_xticklabels(names)
ax.set_ylabel('Score')
ax.set_title('Precision & Recall Comparison', fontsize=13)
ax.legend()
ax.set_ylim(0, 1)
plt.tight_layout()
plt.show()
```

### 6.3 Confusion Matrices

```python
# --- 6.3 Confusion Matrices ---
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes_flat = axes.flatten()

for i, (name, model) in enumerate(models_dict.items()):
    cm = confusion_matrix(y_test, results[name]['y_pred'])
    disp = ConfusionMatrixDisplay(cm, display_labels=['No', 'Yes'])
    disp.plot(ax=axes_flat[i], cmap='Blues', colorbar=False)
    axes_flat[i].set_title(name, fontsize=12)

# Hide the 6th (empty) subplot
axes_flat[5].axis('off')
fig.suptitle('Confusion Matrices for All Models', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

### 6.4 ROC Curves

```python
# --- 6.4 ROC Curves ---
plt.figure(figsize=(8, 6))
colors = ['#3498db','#2ecc71','#e74c3c','#f39c12','#9b59b6']

for (name, res), color in zip(results.items(), colors):
    fpr, tpr, _ = roc_curve(y_test, res['y_prob'])
    plt.plot(fpr, tpr, label=f"{name} (AUC={res['auc']:.4f})", linewidth=2, color=color)

plt.plot([0,1], [0,1], 'k--', linewidth=1, label='Random Classifier')
plt.title('ROC Curves â€“ All Models', fontsize=13)
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 6.5 AUC Score Bar Chart

```python
# --- 6.5 AUC Bar Chart ---
aucs = [results[n]['auc'] for n in names]

plt.figure(figsize=(10, 5))
bars = plt.barh(names, aucs, color=['#3498db','#2ecc71','#e74c3c','#f39c12','#9b59b6'],
                edgecolor='black')
for bar, auc in zip(bars, aucs):
    plt.text(bar.get_width() + 0.005, bar.get_y() + bar.get_height()/2,
             f'{auc:.4f}', va='center', fontweight='bold')
plt.xlim(0.8, 1.0)
plt.title('AUC Scores by Model', fontsize=13)
plt.xlabel('AUC Score')
plt.tight_layout()
plt.show()
```

### 6.6 Summary Table

```python
# --- 6.6 Summary Table ---
summary_df = pd.DataFrame({
    'Accuracy':  {n: r['accuracy']  for n, r in results.items()},
    'Precision': {n: r['precision'] for n, r in results.items()},
    'Recall':    {n: r['recall']    for n, r in results.items()},
    'AUC':       {n: r['auc']       for n, r in results.items()}
}).round(4)

print('ðŸ“Š Model Comparison Summary Table:')
print('=' * 60)
print(summary_df.to_string())
print('=' * 60)
```

## Section 7 â€“ Conclusion

### What the Results Tell Us
- **Neural Network** achieved the best overall performance (Accuracy: ~92.15%, AUC: ~0.9494), demonstrating its ability to capture complex, non-linear decision boundaries in the data.
- **Decision Tree** and **Logistic Regression** performed closely behind, offering good trade-offs between interpretability and performance.
- **Naive Bayes** had the lowest accuracy but the **highest recall** â€” making it useful when the cost of missing a potential subscriber is high.
- **KNN** performed reasonably but is more sensitive to class imbalance and high-dimensional feature spaces.

### Why These Results Occur
- The dataset is heavily imbalanced (88.7% "no"), which inherently suppresses recall for the minority "yes" class across all models.
- `duration` (call length) is the most predictive feature, but it is only known *after* the call ends â€” a significant real-world limitation.
- Economic features like `euribor3m` and `emp.var.rate` provide macro-level context that helps models generalize beyond individual client attributes.
- Neural Networks capture non-linear feature interactions better than linear models like Logistic Regression, which explains the MLP's superior AUC.

### Challenges Faced
- **Class imbalance** makes raw accuracy misleading â€” AUC and Recall are more informative evaluation metrics.
- **Multicollinearity** among economic indicators (`emp.var.rate`, `nr.employed`, `euribor3m`) introduces redundancy.
- **K-Means clusters** do not perfectly align with binary labels since clustering is unsupervised â€” the PCA projection reveals overlapping groups.
- Choosing the right K for K-Means required the **Elbow Method**, and the "elbow" is not always sharply defined.
